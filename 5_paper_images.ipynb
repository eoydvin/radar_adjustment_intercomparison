```python
## prepare cdf data
# get hourly data
rain_gauges_orr = ds_gauges_openrainer.rainfall_amount.values.flatten()
rain_rad_along_cml_orr = ds_rad_along_cml_openrainer.rainfall_amount.resample(time='1h').sum().values.flatten()
rain_radar_orr = ds_radar_openrainer.rainfall_amount.resample(time='1h').sum().values.flatten()

rain_gauges_omrg = ds_gauges_openmrg.rainfall_amount.values.flatten()
rain_rad_along_cml_omrg = ds_rad_along_cml_openmrg.rainfall_amount.resample(time='1h').sum().values.flatten()
rain_radar_omrg = ds_radar_openmrg.rainfall_amount.resample(time='1h').sum().values.flatten()

# remove nan and values below 0.1 mm/h
threshold = 0.1
rain_gauges_orr = rain_gauges_orr[(~np.isnan(rain_gauges_orr)) & (rain_gauges_orr >= threshold)]
rain_rad_along_cml_orr = rain_rad_along_cml_orr[(~np.isnan(rain_rad_along_cml_orr)) & (rain_rad_along_cml_orr >= threshold)]
rain_radar_orr = rain_radar_orr[(~np.isnan(rain_radar_orr)) & (rain_radar_orr >= threshold)]

rain_gauges_omrg = rain_gauges_omrg[(~np.isnan(rain_gauges_omrg)) & (rain_gauges_omrg >= threshold)]
rain_rad_along_cml_omrg = rain_rad_along_cml_omrg[(~np.isnan(rain_rad_along_cml_omrg)) & (rain_rad_along_cml_omrg >= threshold)]
rain_radar_omrg = rain_radar_omrg[(~np.isnan(rain_radar_omrg)) & (rain_radar_omrg >= threshold)]


def get_cdf(data):
    sorted_data = np.sort(data)
    cdf = np.linspace(0, 1, len(sorted_data))
    return sorted_data, cdf

x_g_orr, cdf_g_orr = get_cdf(rain_gauges_orr)
x_rc_orr, cdf_rc_orr = get_cdf(rain_rad_along_cml_orr)
x_r_orr, cdf_r_orr = get_cdf(rain_radar_orr)

x_g_omrg, cdf_g_omrg = get_cdf(rain_gauges_omrg)
x_rc_omrg, cdf_rc_omrg = get_cdf(rain_rad_along_cml_omrg)
x_r_omrg, cdf_r_omrg = get_cdf(rain_radar_omrg)
```

# VARIOGRAM CHECK ANALYSIS

```python
# Load data unchecked

metrics = ['root_mean_square_error', 'percent_bias',
           'mean_absolute_error','pearson_correlation_coefficient']
 
# Define hatches for statuses
hatch_map = {
    'vandebeek': '\\\\',
    'variogram': '-', 
    'variogram_fixnugget0': '/',
    'variogram_fixnugget03': '///',
    'variogram_fixnugget06': '//////',
}

def extract_method_name(path):
    filename = path.split('/')[-1].split('.')[0]
    parts = filename.split('_')
    if len(parts) >= 3 and parts[1] == 'ked':
        return '_'.join(parts[1:3])   # e.g., ked_b
    elif len(parts) >= 4 and (parts[1] == 'add' or parts[1] == 'mul'):
        return '_'.join(parts[1:4])   # e.g., mul_b_ok or add_b_test
    else:
        return None
        
def extract_status(filepath):
    # Get the filename without the path and extension
    filename = filepath.split('/')[-1].split('.')[0]
    # Split into parts by underscore
    parts = filename.split('_')
    # Remove 'est' if present
    cleaned = [p for p in parts if p != 'est']
    # Your extraction logic
    if not cleaned:
        return None
    if len(cleaned) >= 2 and cleaned[-2] == 'variogram' and cleaned[-1].startswith('fixnugget'):
        return f"{cleaned[-2]}_{cleaned[-1]}"
    if cleaned[-1] == 'vandebeek':
        return 'vandebeek'
    if cleaned[-1] == 'variogram':
        return 'variogram'
    return None
    
fig, ax = plt.subplots(2, 1, figsize=(5, 3), sharex=True)

# Load data 
statuses = ['vandebeek', 'variogram', 'variogram_fixnugget0', 'variogram_fixnugget03', 'variogram_fixnugget06']
methods = ['add_p_ok', 'add_b_ok', 'mul_p_ok', 'mul_b_ok', 'ked_p', 'ked_b']
folder = './data/metrics/'
for k, dataset in enumerate(['OpenMRG', 'OpenRainER']):
    data_read = []
    for method in methods:
        for status in statuses:
            file = folder + dataset + '_' + method + '_' + status + '.csv'
            data_read.append(pd.read_csv(file))

    data = pd.concat(data_read)
    
    meth_ = data['method'].apply(extract_method_name)
    stat_ = data['method'].apply(extract_status)
    
    data['method'] = meth_
    data['status'] = stat_

    width = 0.1
    x = np.arange(len(methods))

    for j, status in enumerate(statuses):
        values = []
        for method in methods:
            # root_mean_square_error, mean_absolute_error, pearson_correlation_coefficient
            val = data[(data['method'] == method) & (data['status'] == status)]['root_mean_square_error'][0] 
            values.append(val)
            
        values = np.array(values)

        bars = ax[k].bar(x + j*width - 0.1, values, width=width, edgecolor='black',
                      color=[palette.get(method) for method in list(methods)])
        
        # Apply hatches
        for bar in bars:
            bar.set_hatch(hatch_map[status])
    
        ax[k].set_xticks(x + width)
        ax[k].set_xticklabels(list(methods), rotation=0, ha='center', fontsize=7)
        ax[k].set_ylabel(dataset, fontsize=8)
        ax[k].tick_params(axis='y', labelsize=7)
        ax[k].grid(linestyle='--', color='gray', axis='y')
        ax[k].set_axisbelow(True)

ax[0].set_ylabel('RMSE \n OpenMRG')
ax[1].set_ylabel('RMSE \n OpenRainER')

plt.tight_layout()
fig.savefig('./figures/variogramcheck_analysis.png', dpi=600, bbox_inches='tight')
```
